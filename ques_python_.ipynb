```python
# sort a list in ascending order without using sorted or sort-->(BUBBLE SORT-)
l=eval(input("enter your list"))
print(l)
for i in range(0,len(l)):
    for j in range(i,len(l)):
        if l[i]>l[j]:
           temp=l[j]
           l[j]=l[i]
           l[i]=temp
print(f"sorted list of given list is {l}")
```

```python
# sort a list in descending order without using sorted or sort-->(BUBBLE SORT-)
l=eval(input("enter your list"))
print(l)
for i in range(0,len(l)):
    for j in range(i,len(l)):
        if l[i]<l[j]:
           temp=l[j]
           l[j]=l[i]
           l[i]=temp
print(f"sorted list of given list is {l}")
```
